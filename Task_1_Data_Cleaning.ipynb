# Task 1 – Data Cleaning and Preprocessing

## Customer Personality Analysis Dataset

This notebook demonstrates data cleaning and preprocessing using Python and Pandas.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("raw_dataset.csv", sep="\t")

print("Dataset loaded successfully!")
```

```python
print("First 5 rows:")
df.head()
```

```python
print("Number of rows and columns:")
print(df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())
```

```python
print("Data types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Cleaned column names:")
print(df.columns.tolist())
```

```python
print(df.isnull().sum())
```

```python
df["income"] = pd.to_numeric(df["income"], errors="coerce")
```

```python
median_income = df["income"].median()

df["income"] = df["income"].fillna(median_income)

print("Missing income values after cleaning:")
print(df["income"].isnull().sum())
```

```python
df["education"] = (
    df["education"]
    .astype(str)
    .str.strip()
    .str.title()
)

df["marital_status"] = (
    df["marital_status"]
    .astype(str)
    .str.strip()
    .str.title()
)

print("Text values standardized successfully.")
```

```python
print("Education categories:")
print(df["education"].unique())

print("\nMarital Status categories:")
print(df["marital_status"].unique())
```

```python
df["dt_customer"] = pd.to_datetime(
    df["dt_customer"],
    dayfirst=True,
    errors="coerce"
)

print("Date data type:")
print(df["dt_customer"].dtype)
```

```python
print("Duplicate rows:")
print(df.duplicated().sum())
```

```python
df["year_birth"] = pd.to_numeric(
    df["year_birth"],
    errors="coerce"
)

print(df[["year_birth", "income"]].dtypes)
```

```python
print("Income statistics:")
print(df["income"].describe())
```

```python
print("Year of Birth statistics:")
print(df["year_birth"].describe())
```

```python
print("FINAL DATA QUALITY CHECK")
print("------------------------")

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDataset shape:")
print(df.shape)

print("\nData types:")
print(df.dtypes)
```

```python
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
